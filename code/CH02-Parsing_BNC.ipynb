```python
nlp = stanza.Pipeline('en')  # 분석 파이프라인 객체 생성
```

```python
input_file = "bnc_sample.txt"
output_file = input_file.replace(".txt", "_tagged.txt")
```

```python
with open(input_file, "r", encoding="utf-8") as f:
    sentences = [line.strip() for line in f]

with open(output_file, "w", encoding="utf-8") as f:
    for i, sentence in enumerate(sentences):
        print(f"<s id=\"{i+1}\">", file=f)
        doc = nlp(sentence)
        for sent in doc.sentences:
            for word in sent.words:
                print(word.text, word.upos, word.lemma, word.id, word.head,
                      sent.words[word.head-1].text if word.head > 0 else "root",
                      word.deprel, sep="\t", file=f)
        print("</s>", file=f)
```
